```python
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download required resources
nltk.download('stopwords')

# Input text document
text = """
Natural Language Processing is an interesting field.
We are learning text preprocessing techniques.
The students are studying and playing games.
"""

print("Original Text:")
print(text)

# Step 1: Convert text to lowercase
text = text.lower()

# Step 2: Remove punctuation and special characters
text = re.sub(r'[^a-zA-Z\s]', '', text)

# Step 3: Tokenization (split text into words)
words = text.split()

print("\nTokenized Words:")
print(words)

# Step 4: Remove stop words
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print("\nAfter Stop Word Removal:")
print(filtered_words)

# Step 5: Apply stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\nAfter Stemming:")
print(stemmed_words)

# Final preprocessed text
final_text = " ".join(stemmed_words)

print("\nFinal Preprocessed Text:")
print(final_text)
```
