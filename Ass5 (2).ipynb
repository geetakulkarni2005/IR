```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin

url = "https://books.toscrape.com/"

response = requests.get(
    url,
    timeout=20,
    headers={"User-Agent": "Mozilla/5.0"}
)

print("Status Code:", response.status_code)

soup = BeautifulSoup(response.text, "html.parser")

print("Website title:", soup.title.get_text(strip=True))
```

```python
products = []

# Find all product containers
items = soup.select("article.product_pod")

for item in items:
    # Product name
    name = item.h3.a["title"]

    # Product price
    price = item.select_one(".price_color").get_text(strip=True)

    # Product availability
    availability = item.select_one(
        ".availability"
    ).get_text(" ", strip=True)

    # Product link
    relative_link = item.h3.a["href"]
    product_link = urljoin(url, relative_link)

    # Store product details
    products.append({
        "Product Name": name,
        "Price": price,
        "Availability": availability,
        "Product Link": product_link
    })

# Convert to DataFrame
df = pd.DataFrame(products)

print(df.head(10))
print("\nTotal products extracted:", len(df))
```

```python
for index, row in df.iterrows():
    print("Product:", row["Product Name"])
    print("Price:", row["Price"])
    print("Link:", row["Product Link"])
    print("-" * 60)
```
