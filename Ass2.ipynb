```python
# Document Retrieval Using Inverted Files

import re
from collections import defaultdict

# Sample documents
documents = {
    1: "Python is a programming language",
    2: "Python is used for machine learning",
    3: "Machine learning uses data science",
    4: "Information retrieval searches documents"
}

# Step 1: Create an inverted index
inverted_index = defaultdict(list)

for doc_id, text in documents.items():
    words = re.findall(r'\b\w+\b', text.lower())

    for word in set(words):
        inverted_index[word].append(doc_id)

# Display the inverted index
print("Inverted Index:")
for word, doc_ids in sorted(inverted_index.items()):
    print(word, ":", doc_ids)


# Step 2: Retrieve documents using a search query
query = input("\nEnter search query: ").lower()
query_words = re.findall(r'\b\w+\b', query)

# Find documents containing the query words
result_docs = set()

for word in query_words:
    if word in inverted_index:
        result_docs.update(inverted_index[word])

# Display retrieved documents
print("\nRetrieved Documents:")

if result_docs:
    for doc_id in sorted(result_docs):
        print("Document", doc_id, ":", documents[doc_id])
else:
    print("No documents found.")
```
