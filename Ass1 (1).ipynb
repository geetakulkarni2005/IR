```python
text = input("Enter text:")
text = text.lower()
words = text.split()
print("\nOriginal  words:", words)

stop_words = ["and","the","is","on","a","an","off"]
filtered = [w for w in words if w not in stop_words]
print("\nAfter stop_words:", filtered)

stemmed = []

for word in filtered:
    if word.endswith("ing"):
        word = word[:-3]
    elif word.endswith("s"):
        word = word[:-1]

    stemmed.append(word)

print("\nAfter stemming:", stemmed)
print("\nFinal text:", " ".join(stemmed))
```
