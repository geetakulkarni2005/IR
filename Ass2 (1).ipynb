```python
documents = {
    1: "python is easy",
    2: "java is easy",
    3: "python is powerful"
}

index ={}

for doc_id, text in documents.items():
    words = text.split()

    for word in words:
        if word not in index:
            index[word] = []

        index[word].append(doc_id)

print("Inverted index:")
print(index)

search = input("\nEnter word to search:").lower()

if search in index:
    print("Documents Found:", index[search])

    for doc_id in index[search]:
        print("Document", doc_id, ":", documents[doc_id])

else:
    print("Document Not Found")
```
